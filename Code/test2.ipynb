```python
import gurobipy as gp
from gurobipy import GRB
import random

# Create model
m = gp.Model("knapsack")

# Problem size (increase to make it harder)
n_items = 2000

# Random weights and values
random.seed(42)
weights = [random.randint(1, 50) for _ in range(n_items)]
values  = [random.randint(10, 100) for _ in range(n_items)]
capacity = int(0.3 * sum(weights))

# Binary decision variables
x = m.addVars(n_items, vtype=GRB.BINARY, name="x")

# Objective: maximize total value
m.setObjective(gp.quicksum(values[i] * x[i] for i in range(n_items)), GRB.MAXIMIZE)

# Capacity constraint
m.addConstr(gp.quicksum(weights[i] * x[i] for i in range(n_items)) <= capacity)

# Optional: make it visibly work
m.setParam("TimeLimit", 30)      # seconds
m.setParam("MIPFocus", 1)        # focus on finding incumbents
m.setParam("Threads", 1)         # deterministic logs
m.setParam("LogFile", "Output/Mathematical/solve.log")

m.optimize()

m.write("Output/Mathematical/model.lp")     # human-readable model
m.write("Output/Mathematical/model.mps")    # solver-standard format
m.write("Output/Mathematical/solution.sol")
```
